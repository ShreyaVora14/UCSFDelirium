```python
import pandas as pd
from imblearn.over_sampling import SMOTE
from collections import Counter
```

```python
X_oversample = pd.read_csv('/Users/shreyavora/Desktop/deliriumPaper/UCSFDelirium/Serial Transciptomics/WithoutSMOTE/serialGenes.csv')
X_oversample.head()
```

```python
y_oversample = X_oversample['Delirium']
#X = X.drop('Diagnosis', axis=1)  # Keep 'Steroids'
X_oversample = X_oversample.drop('Delirium', axis=1)
X_oversample.columns = X_oversample.columns.astype(str)
```

```python
oversample = SMOTE()
counter = Counter(y_oversample)
print(counter)
```

```python
X_oversampled, y_oversampled = oversample.fit_resample(X_oversample, y_oversample)
# summarize the new class distribution
# scatter plot of examples by class label
X_oversampled.shapecounter = Counter(y_oversampled)
print(Counter(y_oversampled))
```

```python
df_resampled = pd.concat([X_oversampled, y_oversampled], axis=1)
df_resampled.to_csv('oversampled_serial.csv', index=False)
```
